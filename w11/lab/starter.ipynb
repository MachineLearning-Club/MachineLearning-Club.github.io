# Week 11 — a small, testable classroom monitor

Run the cells in order. Before each run, predict the output. Change one input, then explain what changed. No API, key, installation, or personal information is needed.


## 1. Name the data and the required behaviour


```python
examples=[(17,"HEAT ON"),(21,"HEAT OFF"),(None,"CHECK SENSOR")]
for value,expected in examples:
    print("Input",value,"must produce",expected)
```

## 2. Keep the sensor separate from a decision


```python
class Sensor:
    def __init__(self,value):self.value=value
    def read(self):return self.value

def decide(value):
    if value is None:return "CHECK SENSOR"
    return "HEAT ON" if value<19 else "HEAT OFF"

for value,expected in examples:
    result=decide(Sensor(value).read())
    print(value,"=>",result)
    assert result==expected
```

## 3. Test the exact boundary


```python
print("18 degrees:",decide(18))
print("19 degrees:",decide(19))
print("No reading:",decide(None))
assert decide(18)=="HEAT ON"
assert decide(19)=="HEAT OFF"
assert decide(None)=="CHECK SENSOR"
```
